```python
# Starting populations
A = float(input("Enter population of City A: "))
B = float(input("Enter population of City B: "))
C = float(input("Enter population of City C: "))

# Number of years
years = int(input("Enter number of years: "))

P0 = np.array([A, B, C])

print("\nInitial population:")
print(P0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("City Population Migration Model")
print("--------------------------------")
```

```python
M = np.array([
    [0.90, 0.10, 0.05],
    [0.05, 0.85, 0.10],
    [0.05, 0.05, 0.85]
])

print("Migration matrix M:")
print(M)
```

```python
populations = [P0]
P = P0.copy()

for year in range(1, years + 1):
    P = M @ P
    populations.append(P)

populations = np.array(populations)

print("\nPopulation by year:")
for year, pop in enumerate(populations):
    print(f"Year {year}: A={pop[0]:.2f}, B={pop[1]:.2f}, C={pop[2]:.2f}")
```

```python
Mn = np.linalg.matrix_power(M, years)

print("\nM^n:")
print(Mn)

final_population = Mn @ P0

print("\nFinal population:")
print(final_population)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(M)

D = np.diag(eigenvalues)
P = eigenvectors
P_inv = np.linalg.inv(P)

print("Eigenvalues:")
print(eigenvalues)

print("\nMatrix P:")
print(P)

print("\nDiagonal matrix D:")
print(D)

print("\nP inverse:")
print(P_inv)
```

```python
M_from_diagonalization = P @ D @ P_inv

print("M reconstructed using P D P^-1:")
print(M_from_diagonalization)

print("\nOriginal M:")
print(M)
```

```python
D_power = np.linalg.matrix_power(D, years)

Mn_diagonal = P @ D_power @ P_inv

print("M^n using diagonalization:")
print(Mn_diagonal)

print("\nM^n using direct matrix power:")
print(Mn)

print("\nAre they approximately equal?")
print(np.allclose(Mn, Mn_diagonal))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(populations[:, 0], marker='o', label='City A')
plt.plot(populations[:, 1], marker='o', label='City B')
plt.plot(populations[:, 2], marker='o', label='City C')

plt.xlabel("Year")
plt.ylabel("Population")
plt.title("City Population Migration Model")

plt.legend()
plt.grid(True)

plt.show()
```
